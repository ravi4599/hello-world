```python
#data and visuals
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
bain = pd.read_csv("../raw/bain_training_set_raw.csv")
bain.columns = bain.columns.str.replace("vv_bain_acxiom_match_data.", "", 1)
bain.head()
```

```python
print("row count:",bain.shape)
print("unique surveyid count:", np.unique(bain.surveyid).shape)
#bain_oneper = bain.groupby('id').first()
```

```python
duplicateRowsDF = bain[bain.duplicated(['surveyid'])]
duplicateRowsDF.sort_values('surveyid').head()
```

```python
bain_nodup = bain.drop_duplicates('surveyid')
bain_nodup = bain_nodup.drop_duplicates('client_id')
print("row count:",bain_nodup.shape)
print("unique surveyid count:", np.unique(bain_nodup.surveyid).shape)
```

```python
#leaving survey_id in
survey_coluns = ["surveygroupid","sitename","surveydate","userhostaddress","redflag","scored","q126_gender",
                 "q127_age","q131_country_of_residence","q132_zip/postal_code","q128_marital_status",
                 "q129_number_of_children_under_18","q130_annual_household_income","q50_access_url","customer_id",
                 "group_type","q51_access_method","q100","q101","q102","q103","q104","q105","q106","q107","q108","q109",
                 "q110","q111","q112","q113","q114","q115","q116","q117","q118","q119","q120","q121","q122","q123","q124",
                 "q125","q200","assigned_segment","group","id"]

vv_non_acxiom_attributes = ["document_id","client_id","crm_id","c360_id","dob","country","seaware_id","master_id",
                            "vxp_id","data_loaded_from_nl"]

useless_acxiom_attributes = ['person_abilitec_consumerlink', 'place_abilitec_addresslink', 'household_abilitec_householdlink',
                            'place_placeid', 'household_household_id']

bain_acxiom_master = bain_nodup.drop(survey_coluns, axis = 1)
bain_acxiom_master = bain_acxiom_master.drop(vv_non_acxiom_attributes, axis = 1)
bain_acxiom_master = bain_acxiom_master.drop(useless_acxiom_attributes, axis = 1)
```
